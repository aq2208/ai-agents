```python
# response --> payload
import json
import re
from html.parser import HTMLParser


class HTMLTextExtractor(HTMLParser):
    def __init__(self):
        super().__init__()
        self.text_parts = []

    def handle_data(self, data):
        self.text_parts.append(data)

    def get_text(self):
        return ''.join(self.text_parts)


def html_to_text(html: str) -> str:
    parser = HTMLTextExtractor()
    parser.feed(html)
    return parser.get_text().strip()


def map_ticket_to_payload(ticket: dict) -> dict:
    desc = ticket.get("description_text", "")

    # Parse "+ Key: Value" lines from description
    meta = {}
    for line in desc.split("  "):
        line = line.strip()
        if line.startswith("+ ") and ": " in line:
            key, _, value = line[2:].partition(": ")
            meta[key.strip()] = value.strip()

    # Parse nested "Thông tin thêm" sub-fields
    thong_tin_them = {}
    raw_extra = meta.get("Thông tin thêm", "")
    for part in raw_extra.split(" - "):
        part = part.strip().lstrip("-").strip()
        if ": " in part:
            k, _, v = part.partition(": ")
            thong_tin_them[k.strip()] = v.strip()
    if thong_tin_them:
        meta["Thông tin thêm"] = thong_tin_them

    # Map conversations → comments (skip internal notes)
    # category = 1 -> user
    # category = 3 -> CS
    # user_id = 43083493613 -> AI
    CATEGORY_ROLE = {1: "user", 3: "admin"}
    AI_USER_ID = 43083493613

    comments = []
    for conv in ticket.get("conversations", []):
        role = CATEGORY_ROLE.get(conv.get("category"))
        if not role or conv.get('private'):
            continue

        if conv.get("user_id") == AI_USER_ID:
            role = "AI"

        content = conv.get("body_text") or html_to_text(conv.get("body", ""))

        comments.append({
            "role": role,
            "content": content,
            "timestamp": conv.get("created_at"),
        })

    # meta["comments"] = comments

    custom = ticket.get("custom_fields", {})

    payload = {
        "transid": custom.get("cf_transid") or meta.get("TransID", ""),
        "userid": custom.get("cf_userid") or meta.get("UserID", ""),
        "title": ticket.get("subject", ""),
        "freshdesk_id": str(ticket.get("id", "")) + '01',
        "meta": meta,
        "comments": comments
    }

    return payload

```

```python
import requests
from requests.auth import HTTPBasicAuth
import time

api_key = "Otm4mt5dMrvZZ3m5ut"
domain = "vngzalopay"
ticket_id = need_to_resolve

response_list = {}

for i in ticket_id:
    url = f"https://{domain}.freshdesk.com/api/v2/tickets/{i}?include=conversations"
    response = requests.get(
        url,
        auth=HTTPBasicAuth(api_key, "X"),
        headers={"Content-Type": "application/json"}
    )
    time.sleep(1)
    
    response_list[i] = map_ticket_to_payload(response.json())
```
